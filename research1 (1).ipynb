```python
import cv2
import os

video_folder="/content/drive/MyDrive/research_data/FaceForensics++_C23/original"

output_folder="/content/drive/MyDrive/research_data/extracted_frames/real"

videos=os.listdir(video_folder)

frame_count=10

for video in videos:

    cap=cv2.VideoCapture(os.path.join(video_folder,video))

    total=int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

    interval=max(total//frame_count,1)

    count=0
    saved=0

    while cap.isOpened():

        ret,frame=cap.read()

        if not ret:
            break

        if count%interval==0:

            filename=f"{video[:-4]}_{saved}.jpg"

            cv2.imwrite(os.path.join(output_folder,filename),frame)

            saved+=1

            if saved==frame_count:
                break

        count+=1

    cap.release()

print("Original Videos Completed")
```

```python
import cv2
import os

video_folder = "/content/drive/MyDrive/research_data/FaceForensics++_C23/original"
output_folder = "/content/drive/MyDrive/research_data/extracted_frames/real"

videos = sorted(os.listdir(video_folder))

# Test on first 100 videos
videos = videos[:100]

frames_per_video = 10

total_saved = 0

for video in videos:

    video_path = os.path.join(video_folder, video)

    cap = cv2.VideoCapture(video_path)

    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

    if total_frames == 0:
        cap.release()
        continue

    step = max(total_frames // frames_per_video, 1)

    frame_id = 0
    saved = 0

    while True:

        ret, frame = cap.read()

        if not ret:
            break

        if frame_id % step == 0:

            filename = f"{video[:-4]}_{saved}.jpg"

            cv2.imwrite(
                os.path.join(output_folder, filename),
                frame
            )

            saved += 1
            total_saved += 1

            if saved == frames_per_video:
                break

        frame_id += 1

    cap.release()

print("Finished")
print("Images Saved:", total_saved)
```

```python
import cv2
import os
from tqdm import tqdm
```

```python
def extract_frames(video_folder,
                   output_folder,
                   num_videos,
                   frames_per_video=10):

    os.makedirs(output_folder, exist_ok=True)

    videos = sorted(os.listdir(video_folder))[:num_videos]

    total_saved = 0

    for video in tqdm(videos):

        video_path = os.path.join(video_folder, video)

        cap = cv2.VideoCapture(video_path)

        total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

        if total_frames == 0:
            cap.release()
            continue

        step = max(total_frames // frames_per_video, 1)

        frame_id = 0
        saved = 0

        while True:

            ret, frame = cap.read()

            if not ret:
                break

            if frame_id % step == 0:

                filename = f"{os.path.basename(video_folder)}_{video[:-4]}_{saved}.jpg"

                cv2.imwrite(
                    os.path.join(output_folder, filename),
                    frame
                )

                saved += 1
                total_saved += 1

                if saved == frames_per_video:
                    break

            frame_id += 1

        cap.release()

    print(f"\nSaved {total_saved} images in {output_folder}")
```

```python
BASE = "/content/drive/MyDrive/research_data/FaceForensics++_C23"
OUT  = "/content/drive/MyDrive/research_data/final_dataset"
```

```python
extract_frames(
    video_folder=os.path.join(BASE, "original"),
    output_folder=os.path.join(OUT, "real"),
    num_videos=500,
    frames_per_video=10
)
```

```python
fake_folders = [
    "Deepfakes",
    "FaceSwap",
    "Face2Face",
    "FaceShifter",
    "NeuralTextures",
    "DeepFakeDetection"
]

for folder in fake_folders:

    print(f"\nProcessing {folder}")

    extract_frames(
        video_folder=os.path.join(BASE, folder),
        output_folder=os.path.join(OUT, "fake"),
        num_videos=100,
        frames_per_video=10
    )
```
